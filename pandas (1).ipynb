```python
import pandas as pd
df = pd.read_csv("Employee.csv")

def performance_category(age):
  if age >30:
    return "Experienced"
  else:
    return "learner"
df["experienced status"] = df["Age"].apply(performance_category)
df.head()
```

```python
city_codes ={"Banglore":"BLRU","New Delhi":"NDLS"}
df["city_codes"]=df["City"].map(city_codes)
```
